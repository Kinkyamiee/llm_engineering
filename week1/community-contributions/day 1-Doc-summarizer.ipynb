```python
def messages_for(reseach_paper):
    return [{"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt + research_paper}
    ]
```

```python
messages_for(research_paper)
```

```python
def summarize(research_paper):
    research = research_paper
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(research_paper)
    )
    return response.choices[0].message.content
```

```python
summarize(research_paper)
```

```python
def display_summary(research_paper):
    summary = summarize(research_paper)
    display(Markdown(summary))
```
